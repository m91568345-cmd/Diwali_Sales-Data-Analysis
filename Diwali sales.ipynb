# EDA (Exploratory Data Analysis)

```python
# Create a barplot basis on gender and it's count 
plt.figure(figsize=(3,4))
ax=sns.countplot(x='Gender',data=df)
for bar in ax.containers:
    ax.bar_label(bar)
plt.show()
```

```python
# we have learn from above graph most number of females customer than male 
```

# create a barplot basis on Gender and total_amount 

```python
total_amount = df.groupby('Gender')['Amount'].sum().reset_index()
sns.barplot(x='Gender',y='Amount',data =total_amount)
plt.show()
```

```python
# from above graph we can see most buyer females then male for purchasing  
```

# Age 

```python
plt.figure(figsize =(5,4))
ax=sns.countplot(x='Age Group', data = df, hue = 'Gender')
for bar in ax.containers:
    ax.bar_label(bar)
plt.show()
```

```python
# from above graph we can see most number of buyer are coming 26 to 35  age group and female buyers are more than male 
```

# Show total_Amount basis of Age Group 

```python
Age_total = df.groupby('Age Group')['Amount'].sum().reset_index().sort_values(by='Amount',ascending= False)
sns.barplot(x='Age Group',y='Amount',data = Age_total,palette='viridis')
plt.show()
```

From above graphs we can see that most of the buyers are of age group between 26-35 yrs female

# State

Top 10 state basis on total_Amount

```python
plt.figure(figsize=(14,5))
State_Amount = df.groupby('State')['Amount'].sum().reset_index().sort_values(by='Amount',ascending=False).head(10)
plt.xticks(rotation=45)
sns.barplot(x='State',y ='Amount',data =State_Amount,palette = 'viridis',hue='State')
plt.xlabel('State')
plt.ylabel('Amount')
plt.title('State Vs Total_Amount')

plt.show()
```

Top 10 state basis on total_Amount

```python
plt.figure(figsize=(14,5))
State_Amount = df.groupby('State')['Orders'].sum().reset_index().sort_values(by='Orders',ascending=False).head(10)
plt.xticks(rotation=45)
sns.barplot(x='State',y ='Orders',data =State_Amount,palette = 'viridis',hue='State')

plt.xlabel('State')
plt.ylabel('Orders')
plt.title('Orders Vs Total_Amount')

plt.show()
```

From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively

# Marital_Status

```python
ax=sns.countplot(x='Marital_Status',data = df,palette='magma',hue = 'Marital_Status')
for bar in ax.containers:
    ax.bar_label(bar)
plt.show()
```

```python
Sales_state = df.groupby(['Marital_Status','Gender'])['Amount'].sum().reset_index().sort_values(by='Amount',ascending=False)
sns.barplot(x='Marital_Status',y='Amount',data =Sales_state,hue='Gender')
plt.xlabel('Marital_Status')
plt.ylabel('Amount')
plt.title('Marital_status Vs Amount')
plt.show()
```

From above graphs we can see that most of the buyers are married (women) and they have high purchasing power

# Occupation 

```python
plt.figure(figsize=(18,5))
ax=sns.countplot(x='Occupation',data = df,palette='magma',hue = 'Occupation')
for bar in ax.containers:
    ax.bar_label(bar)
plt.xticks(rotation=45)
plt.show()
```

from above graph , we can seen most emp's occupation is (IT,Healthcare,Aviation,Banking)

 # Occupation basis on Amount 

```python
plt.figure(figsize=(12,5))
occupation_amount = df.groupby('Occupation')['Amount'].sum().reset_index().sort_values(by='Amount',ascending=False)

sns.barplot(x='Occupation',y ='Amount',data = occupation_amount,)

plt.title('Occupation Vs Total_Amount')

plt.xticks(rotation=45)
plt.show()
```

From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector

#  Product_Category

```python
plt.figure(figsize=(12,4))
ax=sns.countplot(x='Product_Category',data = df)
for bar in ax.containers:
    ax.bar_label(bar)
plt.xticks(rotation=45)
plt.show()
```

From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category

```python
Product_Amount=df.groupby('Product_Category')['Amount'].sum().reset_index().sort_values(by='Amount',ascending=False).head(10)
plt.figure(figsize=(12,4))
sns.barplot(x='Product_Category',y='Amount', data = Product_Amount,palette='magma',hue='Product_Category')
plt.xticks(rotation = 45)
plt.show()
```

From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category

# Product_Category Vs Orders 

```python
Product_Amount=df.groupby('Product_Category')['Orders'].sum().reset_index().sort_values(by='Orders',ascending=False).head(10)
plt.figure(figsize=(12,4))
sns.barplot(x='Product_Category',y='Orders', data = Product_Amount,palette='magma',hue='Product_Category')
plt.ylabel("Orders")
plt.xlabel("Product_Category")
plt.xticks(rotation = 45)
plt.show()
```
